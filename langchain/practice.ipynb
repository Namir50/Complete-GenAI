CREATING AGENT

```python
from langchain.agents import create_agent
```

```python
agent = create_agent(model='google_genai:gemini-3.6-flash')
```

```python
from langchain.messages import HumanMessage
```

```python
response = agent.invoke(
    {"messages": [HumanMessage(content = "What is the capital of moon?")]}
)
```

```python
response
```

```python
response['messages'][-1].content
```

```python
print(response['messages'][-1].content)
```

Passing ai messages also to trick the model

```python
from langchain.messages import AIMessage

response = agent.invoke(
    {"messages":[HumanMessage(content="What is the capital of moon?"),
    AIMessage(content="The capital of moon is luna"),
    HumanMessage(content="Interesting, tell me more about luna")]}
)

response
```

Streaming

```python
for token, metadata in agent.stream(
    {"messages": [HumanMessage(content="Tell me about luna, the capital city of moon")]},
    stream_mode="messages"
): 
    if token.content:
        # Use end="" (empty string) instead of end=" "
        print(token.content, end="", flush=True)
```

```python
for token, metadata in agent.stream(
    {"messages": [HumanMessage(content="Tell me about luna, the capital city of moon")]},
    stream_mode="messages"
): 
    if token.content:
        # Direct fix using your previous indexing style
        print(token.content[0]['text'], end="", flush=True)
```

Giving system prompt to make it understand our context

```python
from langchain.agents import create_agent
from langchain.messages import HumanMessage, SystemMessage

agent = create_agent(model = 'google_genai:gemini-3.6-flash')
```

```python
response = agent.invoke(
    {"messages": [SystemMessage(content="You are a science fiction writer, create a capital city at the users request, just give one word answer and at where the capital is exactly located"),
                  HumanMessage(content = "What is the capital of moon?")]}
)
```

```python
response
```

```python
response['messages'][-1].content[0]['text']
```

```python
print(response['messages'][-1].content[0]['text'])
```

Structured output

```python
from langchain.agents import create_agent
from langchain.messages import HumanMessage, SystemMessage
from pydantic import BaseModel
```

```python
class CapitalInfo(BaseModel):
    name:str
    location:str
    temperature:str

agent = create_agent(model = 'google_genai:gemini-3.6-flash',
                     response_format = CapitalInfo)

response = agent.invoke(
    {"messages":[
        SystemMessage(content="You are a science fiction writer, create a capital city at the users request, just give one word answer and at where the capital is exactly located"),
        HumanMessage(content = "What is the capital of moon?")
    ]}
)
response
```

```python
print(response['messages'][-1].content)
```
